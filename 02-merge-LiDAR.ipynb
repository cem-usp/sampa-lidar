```python
import geopandas as gpd
from tqdm import tqdm  # barra de progresso (opcional)
```

```python
RESULT_FOLDER = '/Users/fernandogomes/dev/LiDAR_produtos'
```

```python
gdf_articulacao_17_20 = gpd.read_file("zip://data/SIRGAS_SHP_quadriculamdt.zip!/SIRGAS_SHP_quadriculamdt/")
gdf_articulacao_24 = gpd.read_file('results/folhas_sp_cortada.gpkg')
gdf_distritos = gpd.read_file('data/SIRGAS_GPKG_distrito.gpkg')
```

```python
gdf_articulacao_24.to_crs(epsg=31983, inplace=True)
```

```python
gdf_articulacao_17_20.set_crs(epsg=31983, inplace=True)
```

```python
gdf_distritos.set_crs(epsg=31983, inplace=True)
```

```python
# Garante que todos os GeoDataFrames estão no mesmo CRS
crs = gdf_distritos.crs
gdf_articulacao_17_20 = gdf_articulacao_17_20.to_crs(crs)
gdf_articulacao_24 = gdf_articulacao_24.to_crs(crs)

for _, distrito in tqdm(gdf_distritos.iterrows(), total=len(gdf_distritos)):
    nome_distrito = distrito["ds_nome"].replace(' ', '-')
    cod_distrito = distrito["ds_codigo"]

    folhas_17_20 = gdf_articulacao_17_20[gdf_articulacao_17_20.intersects(distrito.geometry)].copy()
    folhas_24 = gdf_articulacao_24[gdf_articulacao_24.intersects(distrito.geometry)].copy()

    files_2017 = ' '.join([f'{RESULT_FOLDER}/2017/LiDAR/buildings-{y}-2017-50cm.laz' for y in folhas_17_20.qmdt_cod.to_list()])
    files_2020 = ' '.join([f'{RESULT_FOLDER}/2020/LiDAR/buildings-{y}-2020-50cm.laz' for y in folhas_17_20.qmdt_cod.to_list()])
    files_2024 = ' '.join([f'{RESULT_FOLDER}/2024/LiDAR/buildings-{y}-2024-50cm.laz' for y in folhas_24.nome.to_list()])

    !pdal merge {files_2017} {RESULT_FOLDER}/2017/LiDAR-subs/{cod_distrito}-{nome_distrito}-buildings-50cm.laz
    !pdal merge {files_2020} {RESULT_FOLDER}/2020/LiDAR-subs/{cod_distrito}-{nome_distrito}-buildings-50cm.laz
    !pdal merge {files_2024} {RESULT_FOLDER}/2024/LiDAR-subs/{cod_distrito}-{nome_distrito}-buildings-50cm.laz
```
