```python
import geopandas as gpd
from pathlib import Path
import pandas as pd
```

```python
# === CONFIGURAÇÃO ===
# pasta_arquivos = Path("/Users/fernandogomes/Downloads/arquivos_faltantes")  # Pasta com arquivos
pasta_arquivos = Path("../LiDAR-Sampa-2024")  # Pasta com arquivos
caminho_gpkg = Path("results/folhas_faltantes_IGG_2024.gpkg")  # GeoPackage
nome_camada = "folhas_sp_cortada"  # Nome da camada
coluna_nomes = "nome"    # Coluna com os nomes dos arquivos
extensao = ".laz"            # Extensão dos arquivos
saida_csv = Path("results/comparacao_arquivos.csv")  # Caminho do CSV de saída
```

```python
# === LISTA ARQUIVOS DA PASTA ===
arquivos_pasta = {p.stem.removeprefix("SF-23-").removesuffix("_NPC") for p in pasta_arquivos.glob(f"*{extensao}")}
```

```python
arquivos_pasta
```

```python
# === LÊ NOMES DO GPKG ===
gdf = gpd.read_file(caminho_gpkg, layer=nome_camada)
nomes_gpkg = set(gdf[coluna_nomes].astype(str).dropna().unique())
```

```python
nomes_gpkg
```

```python
# === PREPARA RESULTADOS ===
todos_nomes = arquivos_pasta.union(nomes_gpkg)
resultados = []

for nome in sorted(todos_nomes):
    if nome in nomes_gpkg and nome in arquivos_pasta:
        status = "presente_e_ok"
    elif nome in nomes_gpkg:
        status = "esperado_mas_ausente"
    else:
        status = "extra_na_pasta"
    
    resultados.append({
        "nome": nome,
        "status": status,
        "nome_completo": f"{nome}{extensao}"
    })
```

```python
todos_nomes
```

```python
resultados
```

```python
# === SALVA CSV ===
df = pd.DataFrame(resultados)
df.to_csv(saida_csv, index=False, encoding="utf-8")
print(f"\n✅ Resultado salvo em: {saida_csv.resolve()}")
```
